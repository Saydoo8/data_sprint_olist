# INITIAL EXPLORATORY ANALYSIS
# Preliminary observations only — before cleaning and validation

```python
funnel_initial = mql.merge(closed,on="mql_id",how="left")
```

```python
print(mql.shape)
print(closed.shape)
print(funnel_initial.shape)
```

```python
funnel_initial["converted"] = funnel_initial["seller_id"].notna()
```

```python
funnel_initial["converted"].value_counts()
```

```python
conversion_rate_initial = funnel_initial["converted"].mean() * 100
```

```python
conversion_rate_initial
```

```python
print("Initial conversion rate:", round(conversion_rate_initial, 2), "%")
```

```python
#Which marketing channels seem to convert leads into sellers more often?
```

```python
mql["origin"].value_counts()
```

```python
funnel_initial[funnel_initial["converted"] == True]["origin"].value_counts()
```

```python
total_leads_by_channel = mql["origin"].value_counts()

converted_by_channel = funnel_initial[funnel_initial["converted"] == True]["origin"].value_counts()
```

```python
channel_conversion_rate = (converted_by_channel / total_leads_by_channel) * 100
```

```python
channel_conversion_rate.sort_values(ascending=False)
```

Preliminary observation: Lead-to-seller conversion differs across acquisition channels. paid_search, organic_search, and direct_traffic currently show conversion rates above the overall preliminary rate of 10.52%. The unknown category has the highest observed rate, but it cannot yet be interpreted as an actionable acquisition channel. These findings are exploratory and will be validated after data cleaning and seller-performance analysis.
